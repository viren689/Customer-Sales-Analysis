## 5. Data Cleaning and Preparation

```python
# Clean and prepare Sales Dataset

# Convert Date from text to datetime
sales_df["Date"] = pd.to_datetime(sales_df["Date"], errors="coerce")

# Make sure numerical columns have the correct data type
sales_df["Quantity"] = pd.to_numeric(sales_df["Quantity"], errors="coerce")
sales_df["Price"] = pd.to_numeric(sales_df["Price"], errors="coerce")
sales_df["Total_Sales"] = pd.to_numeric(sales_df["Total_Sales"], errors="coerce")

# Remove duplicate records if any
sales_df = sales_df.drop_duplicates()

print("Sales dataset prepared successfully!")
print("Rows:", sales_df.shape[0])
print("Columns:", sales_df.shape[1])
print("Date data type:", sales_df["Date"].dtype)
```

```python
# Clean and prepare Customer Dataset

# Make sure numerical columns have the correct data type
customer_df["Tenure"] = pd.to_numeric(
    customer_df["Tenure"], errors="coerce"
)

customer_df["MonthlyCharges"] = pd.to_numeric(
    customer_df["MonthlyCharges"], errors="coerce"
)

customer_df["TotalCharges"] = pd.to_numeric(
    customer_df["TotalCharges"], errors="coerce"
)

customer_df["SeniorCitizen"] = pd.to_numeric(
    customer_df["SeniorCitizen"], errors="coerce"
)

customer_df["Churn"] = pd.to_numeric(
    customer_df["Churn"], errors="coerce"
)

# Remove duplicate records
customer_df = customer_df.drop_duplicates()

print("Customer dataset prepared successfully!")
print("Rows:", customer_df.shape[0])
print("Columns:", customer_df.shape[1])
```

## 6. Feature Engineering

```python
# Create calculated columns

# Extract date components
sales_df["Year"] = sales_df["Date"].dt.year
sales_df["Month"] = sales_df["Date"].dt.month
sales_df["Month_Name"] = sales_df["Date"].dt.strftime("%B")
sales_df["Quarter"] = sales_df["Date"].dt.quarter

# Calculate sales independently
sales_df["Calculated_Sales"] = sales_df["Quantity"] * sales_df["Price"]

# Check whether calculated sales matches the existing Total_Sales
sales_df["Sales_Check"] = (
    sales_df["Calculated_Sales"] == sales_df["Total_Sales"]
)

print("Calculated columns created successfully!")

display(
    sales_df[
        [
            "Date",
            "Year",
            "Month",
            "Month_Name",
            "Quarter",
            "Quantity",
            "Price",
            "Total_Sales",
            "Calculated_Sales",
            "Sales_Check"
        ]
    ].head()
)
```

```python
# Validate sales calculations

print("Sales calculation validation:")
print(sales_df["Sales_Check"].value_counts())
```

## 7. Monthly Sales Analysis

```python
#  Monthly Sales Analysis

monthly_sales = (
    sales_df
    .groupby(["Year", "Month", "Month_Name"], as_index=False)["Total_Sales"]
    .sum()
    .sort_values(["Year", "Month"])
)

print("Monthly sales summary:")
display(monthly_sales)
```

```python
# Additional monthly sales metrics

monthly_summary = (
    sales_df
    .groupby(["Year", "Month", "Month_Name"])
    .agg(
        Total_Revenue=("Total_Sales", "sum"),
        Total_Quantity=("Quantity", "sum"),
        Average_Sale=("Total_Sales", "mean"),
        Number_of_Transactions=("Total_Sales", "count")
    )
    .reset_index()
    .sort_values(["Year", "Month"])
)

print("Detailed monthly sales summary:")
display(monthly_summary)
```

## 11. Multiple-Condition Filtering

```python
# Filter using AND condition

high_value_sales = sales_df[
    (sales_df["Quantity"] > 5) &
    (sales_df["Total_Sales"] > 1000)
]

print("High-value sales transactions:")
display(high_value_sales)
print("Number of high-value transactions:", len(high_value_sales))
```

```python
# Filter using OR condition

selected_regions = sales_df[
    (sales_df["Region"] == "North") |
    (sales_df["Region"] == "South")
]

print("Sales from North or South regions:")
display(selected_regions)
print("Number of selected transactions:", len(selected_regions))
```

```python
# Combine AND and OR conditions

priority_sales = sales_df[
    (sales_df["Total_Sales"] > 1000) &
    (
        (sales_df["Region"] == "North") |
        (sales_df["Region"] == "South")
    )
]

print("Priority sales transactions:")
display(priority_sales)
print("Number of priority transactions:", len(priority_sales))
```

```python
# Filter sales transactions where BOTH conditions are true:
# Quantity is greater than 5 AND Total_Sales is greater than 1000

high_value_sales = sales_df[
    (sales_df["Quantity"] > 5) &
    (sales_df["Total_Sales"] > 1000)
]

print("High-value sales transactions:")
display(high_value_sales)

print("Number of high-value transactions:", len(high_value_sales))
```

```python
# Filter sales transactions where EITHER the region is North OR South

selected_regions = sales_df[
    (sales_df["Region"] == "North") |
    (sales_df["Region"] == "South")
]

print("Sales from North or South regions:")
display(selected_regions)

print("Number of selected transactions:", len(selected_regions))
```

```python
# Combine AND and OR conditions to identify
# high-value sales from either the North or South region

priority_sales = sales_df[
    (sales_df["Total_Sales"] > 1000) &
    (
        (sales_df["Region"] == "North") |
        (sales_df["Region"] == "South")
    )
]

print("Priority sales transactions:")
display(priority_sales)

print("Number of priority transactions:", len(priority_sales))
```

## 12. String Operations

```python
# Clean and standardize text data in the sales dataset

# Remove extra spaces from product names
sales_df["Product"] = sales_df["Product"].str.strip()

# Standardize product names using title case
sales_df["Product"] = sales_df["Product"].str.title()

# Remove extra spaces from region names
sales_df["Region"] = sales_df["Region"].str.strip()

# Standardize region names using title case
sales_df["Region"] = sales_df["Region"].str.title()

# Remove extra spaces from customer IDs
sales_df["Customer_ID"] = sales_df["Customer_ID"].str.strip()

# Display the cleaned text columns
display(
    sales_df[
        ["Product", "Region", "Customer_ID"]
    ].head(10)
)

print("Text data cleaned and standardized successfully!")
```


## 13. Dataset Merging

```python
# Standardize the customer ID column name
# so both datasets use the same key for merging

customer_df = customer_df.rename(
    columns={"CustomerID": "Customer_ID"}
)

# Check how many customer IDs are common between both datasets

common_customer_ids = set(sales_df["Customer_ID"]).intersection(
    set(customer_df["Customer_ID"])
)

print("Customer IDs in Sales Dataset:", sales_df["Customer_ID"].nunique())
print("Customer IDs in Customer Dataset:", customer_df["Customer_ID"].nunique())
print("Matching Customer IDs:", len(common_customer_ids))
```

```python
# Merge sales data with customer data using Customer_ID
# A left join keeps all sales transactions in the final dataset

merged_df = pd.merge(
    sales_df,
    customer_df,
    on="Customer_ID",
    how="left"
)

print("Datasets merged successfully!")
print("Merged dataset shape:", merged_df.shape)

# Display the first five rows of the merged dataset
display(merged_df.head())
```

```python
# Check whether the merge created missing customer information

print("Missing customer information after merge:")
print(
    merged_df[
        ["Tenure", "MonthlyCharges", "TotalCharges", "Contract", "Churn"]
    ].isnull().sum()
)
```

## 15. Customer Churn and Retention

```python
# Analyze customer churn status
# Churn = 1 means the customer left
# Churn = 0 means the customer stayed

churn_summary = (
    customer_df["Churn"]
    .value_counts()
    .rename(index={0: "Retained", 1: "Churned"})
    .reset_index()
)

churn_summary.columns = ["Customer_Status", "Customer_Count"]

print("Customer Churn Summary:")
display(churn_summary)
```

```python
# Calculate customer retention and churn rates

total_customers = len(customer_df)

churned_customers = (customer_df["Churn"] == 1).sum()
retained_customers = (customer_df["Churn"] == 0).sum()

churn_rate = (churned_customers / total_customers) * 100
retention_rate = (retained_customers / total_customers) * 100

print("Total Customers:", total_customers)
print("Churned Customers:", churned_customers)
print("Retained Customers:", retained_customers)
print("Churn Rate:", round(churn_rate, 2), "%")
print("Retention Rate:", round(retention_rate, 2), "%")
```

## 16. Customer Lifetime Value

```python
# Calculate customer lifetime value from TotalCharges
# Higher TotalCharges indicates greater customer value

customer_df["Customer_Lifetime_Value"] = customer_df["TotalCharges"]

# Divide customers into three value segments using their lifetime value

customer_df["Customer_Segment"] = pd.qcut(
    customer_df["Customer_Lifetime_Value"],
    q=3,
    labels=["Low Value", "Medium Value", "High Value"],
    duplicates="drop"
)

# Summarize the customer segments

customer_segments = (
    customer_df
    .groupby("Customer_Segment", observed=True)
    .agg(
        Number_of_Customers=("Customer_ID", "count"),
        Average_Lifetime_Value=("Customer_Lifetime_Value", "mean"),
        Average_Monthly_Charges=("MonthlyCharges", "mean"),
        Average_Tenure=("Tenure", "mean")
    )
    .reset_index()
)

print("Customer Value Segments:")
display(customer_segments)
```

```python
# Identify the top 10 customers based on lifetime value

top_lifetime_customers = (
    customer_df
    .sort_values("Customer_Lifetime_Value", ascending=False)
    .head(10)
)

print("Top 10 Customers by Lifetime Value:")

display(
    top_lifetime_customers[
        [
            "Customer_ID",
            "Customer_Lifetime_Value",
            "Tenure",
            "MonthlyCharges",
            "Contract",
            "Churn"
        ]
    ]
)
```

## 19. Cross-Selling Analysis

```python
# Find products purchased by each customer
# This helps identify customers who have purchased multiple products.

customer_products = (
    sales_df
    .groupby("Customer_ID")["Product"]
    .agg(lambda products: sorted(set(products)))
    .reset_index(name="Products_Purchased")
)

# Keep only customers who purchased more than one different product

multi_product_customers = customer_products[
    customer_products["Products_Purchased"].apply(len) > 1
].copy()

print("Customers who purchased multiple products:")
display(multi_product_customers.head(10))

print(
    "Number of customers with multiple product purchases:",
    len(multi_product_customers)
)
```

```python
# Create product pairs for customers who purchased multiple products
# Each pair represents a potential cross-selling opportunity.

from itertools import combinations

product_pairs = []

for products in multi_product_customers["Products_Purchased"]:
    for pair in combinations(products, 2):
        product_pairs.append(pair)

product_pair_counts = pd.Series(product_pairs).value_counts()

print("Most common product combinations:")

display(
    product_pair_counts
    .head(10)
    .rename("Number_of_Customers")
    .reset_index(name="Product_Pair")
)
```
